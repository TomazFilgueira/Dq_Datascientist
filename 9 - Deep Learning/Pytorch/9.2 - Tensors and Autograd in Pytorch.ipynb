# 1) PyTorch Tensors and Gradients

A gradient is simply the rate of change of one value with respect to another. While gradients have their roots in calculus, we won't be getting into any of their mathematical details, but rather we'll focus on the practical implications of them.

In neural networks, gradients tell us how much our model's loss changes when we make a small adjustment to each parameter (weight or bias). If a weight has a gradient of `-0.5`, increasing that weight slightly would decrease the model's loss. The loss is how we measure how wrong our model is - it's a single number calculated by a loss function that quantifies the difference between what your model predicted and the actual values. In other words, neural networks can be viewed as an optimization problem where we use these gradients to adjust our parameters and minimize this loss as much as possible

# 2) Creating Tensors and Data Types

**Tensors** are like NumPy arrays but with extra features for building neural networks. They can automatically track mathematical operations (when enabled) to compute gradients for training, and they can run efficiently on GPUs for faster computation. Just like NumPy arrays, tensors store multi-dimensional data and support mathematical operations, but PyTorch adds the gradient tracking capability that makes neural network training possible. In PyTorch, tensors are implemented by the `torch.Tensor class`.

We can create tensors from Python lists or NumPy arrays, or generate them using helper functions. Key creation APIs include the `torch.tensor` [function](https://docs.pytorch.org/docs/stable/generated/torch.tensor.html), `torch.arange` [function](https://docs.pytorch.org/docs/stable/generated/torch.arange.html), `torch.zeros` [function](https://docs.pytorch.org/docs/stable/generated/torch.zeros.html), and `torch.ones` [function](https://docs.pytorch.org/docs/stable/generated/torch.ones.html).
